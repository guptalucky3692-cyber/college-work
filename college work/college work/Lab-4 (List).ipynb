                                       **LIST**
1) A list in Python is a built-in data structure used to store multiple values in a single variable.
2) A list can store different types of data, such as integers, floats, strings, and even other lists.
Properties:
1) Ordered: List elements maintain their insertion order.
2) Mutable: List elements can be changed after creation.
3) Allows Duplicate Elements.
4) Heterogeneous: A list can store different data types.
5) Dynamic: Elements can be added or removed during program execution.
6) Supports Indexing: Each element can be accessed using its index,Indexing starts from 0.
7) Supports Negative Indexing,Supports Slicing,Allows Nested Lists
8) Iterable: List elements can be accessed one by one using a loop.

```python
PRE-DEFINED FUNCTIONS IN PYTHON
                           
# Program 1: Create and display a list
numbers = [10, 20, 30, 40, 50]
print(numbers)

# Program 2: Find the length of a list
numbers = [10, 20, 30, 40, 50]
print(len(numbers))

# Program 3: Access the first and last elements
numbers = [10, 20, 30, 40, 50]
print(numbers[0])
print(numbers[-1])

# Program 4: Access elements using positive indexing
numbers = [10, 20, 30, 40, 50]
print(numbers[0])
print(numbers[2])
print(numbers[4])

# Program 5: Access elements using negative indexing
numbers = [10, 20, 30, 40, 50]
print(numbers[-1])
print(numbers[-2])
print(numbers[-3])

# Program 6: Perform list slicing
numbers = [10, 20, 30, 40, 50]
print(numbers[1:4])
print(numbers[:3])
print(numbers[2:])

# Program 7: Update a list element
numbers = [10, 20, 30, 40, 50]
numbers[2] = 100
print(numbers)

# Program 8: Add an element using append()
numbers = [10, 20, 30, 40, 50]
numbers.append(60)
print(numbers)

# Program 9: Insert an element using insert()
numbers = [10, 20, 30, 40, 50]
numbers.insert(2, 25)
print(numbers)

# Program 10: Add multiple elements using extend()
numbers = [10, 20, 30]
numbers.extend([40, 50, 60])
print(numbers)

# Program 11: Remove an element using remove()
numbers = [10, 20, 30, 40, 50]
numbers.remove(30)
print(numbers)

# Program 12: Remove an element using pop()
numbers = [10, 20, 30, 40, 50]
numbers.pop(2)
print(numbers)

# Program 13: Delete an element using del
numbers = [10, 20, 30, 40, 50]
del numbers[1]
print(numbers)

# Program 14: Clear all elements from a list
numbers = [10, 20, 30, 40, 50]
numbers.clear()
print(numbers)

# Program 15: Find the maximum element
numbers = [10, 50, 30, 20, 40]
print(max(numbers))

# Program 16: Find the minimum element
numbers = [10, 50, 30, 20, 40]
print(min(numbers))

# Program 17: Find the sum of list elements
numbers = [10, 20, 30, 40, 50]
print(sum(numbers))

# Program 18: Calculate the average of list elements
numbers = [10, 20, 30, 40, 50]
average = sum(numbers) / len(numbers)
print(average)

# Program 19: Sort a list in ascending order
numbers = [50, 20, 40, 10, 30]
numbers.sort()
print(numbers)

# Program 20: Sort a list in descending order
numbers = [50, 20, 40, 10, 30]
numbers.sort(reverse=True)
print(numbers)

# Program 21: Reverse a list
numbers = [10, 20, 30, 40, 50]
numbers.reverse()
print(numbers)

# Program 22: Count occurrences of an element
numbers = [10, 20, 10, 30, 10, 40]
print(numbers.count(10))

# Program 23: Find the index of an element
numbers = [10, 20, 30, 40, 50]
print(numbers.index(30))

# Program 24: Copy a list
numbers = [10, 20, 30, 40, 50]
new_numbers = numbers.copy()
print(new_numbers)

# Program 25: Concatenate two lists
numbers1 = [10, 20, 30]
numbers2 = [40, 50, 60]
numbers3 = numbers1 + numbers2
print(numbers3)
```

```python
# Create an empty list
numbers = []

# Take number of elements from user
n = int(input())

# Take elements from user
for i in range(n):
    value = int(input())
    numbers.append(value)

# Display the original list
print(numbers)

# Find the length of the list
print(len(numbers))

# Find the maximum value
print(max(numbers))

# Find the minimum value
print(min(numbers))

# Find the sum of all values
print(sum(numbers))

# Sort the list in ascending order
print(sorted(numbers))

# Sort the list in descending order
print(sorted(numbers, reverse=True))

# Reverse the list
numbers.reverse()
print(numbers)

# Count occurrence of a value
value = int(input())
print(numbers.count(value))

# Find the index of a value
value = int(input())
if value in numbers:
    print(numbers.index(value))

# Insert an element at a particular position
position = int(input())
value = int(input())
numbers.insert(position, value)
print(numbers)

# Remove an element
value = int(input())
if value in numbers:
    numbers.remove(value)
print(numbers)

# Remove the last element
numbers.pop()
print(numbers)
```

```python
# Create an empty list
numbers = []

# Take number of elements
n = int(input())

# Add elements using append()
for i in range(n):
    value = int(input())
    numbers.append(value)

# Display the list
print(numbers)

# Find length using len()
print(len(numbers))

# Find maximum using max()
print(max(numbers))

# Find minimum using min()
print(min(numbers))

# Find sum using sum()
print(sum(numbers))

# Sort using sorted()
print(sorted(numbers))

# Sort in descending order using sorted()
print(sorted(numbers, reverse=True))

# Sort the original list using sort()
numbers.sort()
print(numbers)

# Reverse the original list using reverse()
numbers.reverse()
print(numbers)

# Take a value for searching
value = int(input())

# Check using 'in'
print(value in numbers)

# Check using 'not in'
print(value not in numbers)

# Count occurrence using count()
print(numbers.count(value))

# Find index using index()
if value in numbers:
    print(numbers.index(value))

# Insert an element using insert()
position = int(input())
new_value = int(input())
numbers.insert(position, new_value)
print(numbers)

# Extend the list using extend()
m = int(input())
new_list = []

for i in range(m):
    new_list.append(int(input()))

numbers.extend(new_list)
print(numbers)

# Copy the list using copy()
copy_list = numbers.copy()
print(copy_list)

# Remove an element using remove()
value = int(input())

if value in numbers:
    numbers.remove(value)

print(numbers)

# Remove last element using pop()
if len(numbers) > 0:
    numbers.pop()

print(numbers)

# Delete an element using del
if len(numbers) > 0:
    del numbers[0]

print(numbers)

# Clear the complete list using clear()
numbers.clear()
print(numbers)
```

Application: Student Marks Management System

```python
# Create an empty list
marks = []

# Take number of students
n = int(input())

# Enter marks of students
for i in range(n):
    mark = int(input())
    marks.append(mark)

# Display all marks
print(marks)

# Total number of students
print(len(marks))

# Highest marks
print(max(marks))

# Lowest marks
print(min(marks))

# Total marks
print(sum(marks))

# Average marks
print(sum(marks) / len(marks))

# Marks in ascending order
print(sorted(marks))

# Marks in descending order
print(sorted(marks, reverse=True))

# Sort the original list
marks.sort()
print(marks)

# Reverse the list
marks.reverse()
print(marks)

# Search for a particular mark
search = int(input())
print(search in marks)

# Check whether a mark is not present
print(search not in marks)

# Count how many students obtained the searched mark
print(marks.count(search))

# Find the position of the searched mark
if search in marks:
    print(marks.index(search))

# Add marks of a new student
new_mark = int(input())
marks.append(new_mark)
print(marks)

# Insert marks at a particular position
position = int(input())
mark = int(input())
marks.insert(position, mark)
print(marks)

# Add marks of multiple students
m = int(input())
new_marks = []

for i in range(m):
    new_marks.append(int(input()))

marks.extend(new_marks)
print(marks)

# Create a copy of marks
marks_copy = marks.copy()
print(marks_copy)

# Remove a particular mark
remove_mark = int(input())

if remove_mark in marks:
    marks.remove(remove_mark)

print(marks)

# Remove the last mark
if len(marks) > 0:
    marks.pop()

print(marks)

# Delete the first mark
if len(marks) > 0:
    del marks[0]

print(marks)

# Clear all marks
marks.clear()
print(marks)
```

Tkinter UI-based Student Result Management System

```python
import tkinter as tk
from tkinter import messagebox

# Create main window
root = tk.Tk()
root.title("Student Result Management System")
root.geometry("750x600")

# Create lists
names = []
roll_numbers = []
marks = []


# Add student
def add_student():
    name = name_entry.get()
    roll = roll_entry.get()
    mark = marks_entry.get()

    if name == "" or roll == "" or mark == "":
        messagebox.showwarning("Warning", "Please enter all details")
        return

    names.append(name)
    roll_numbers.append(roll)
    marks.append(int(mark))

    display_list.insert(
        tk.END,
        f"{roll}     {name}     {mark}"
    )

    name_entry.delete(0, tk.END)
    roll_entry.delete(0, tk.END)
    marks_entry.delete(0, tk.END)


# Delete selected student
def delete_student():
    selected = display_list.curselection()

    if not selected:
        messagebox.showwarning("Warning", "Select a student")
        return

    index = selected[0]

    names.pop(index)
    roll_numbers.pop(index)
    marks.pop(index)

    display_list.delete(index)


# Search student
def search_student():
    roll = search_entry.get()

    display_list.delete(0, tk.END)

    for i in range(len(roll_numbers)):
        if roll_numbers[i] == roll:
            display_list.insert(
                tk.END,
                f"{roll_numbers[i]}     {names[i]}     {marks[i]}"
            )


# Sort marks
def sort_marks():
    display_list.delete(0, tk.END)

    sorted_data = sorted(
        zip(marks, roll_numbers, names),
        reverse=True
    )

    for mark, roll, name in sorted_data:
        display_list.insert(
            tk.END,
            f"{roll}     {name}     {mark}"
        )


# Analyze result
def analyze_result():

    if len(marks) == 0:
        messagebox.showwarning("Warning", "No student data")
        return

    total = sum(marks)
    average = total / len(marks)
    maximum = max(marks)
    minimum = min(marks)

    result_label.config(
        text=f"Total: {total}     "
             f"Average: {average:.2f}     "
             f"Maximum: {maximum}     "
             f"Minimum: {minimum}"
    )


# Clear all data
def clear_data():

    names.clear()
    roll_numbers.clear()
    marks.clear()

    display_list.delete(0, tk.END)

    result_label.config(
        text="Total:     Average:     Maximum:     Minimum:"
    )


# ---------------- GUI ----------------

# Heading
title_label = tk.Label(
    root,
    text="STUDENT RESULT MANAGEMENT SYSTEM",
    font=("Arial", 20, "bold")
)
title_label.pack(pady=15)


# Student Name
tk.Label(
    root,
    text="Student Name",
    font=("Arial", 12)
).pack()

name_entry = tk.Entry(
    root,
    width=40,
    font=("Arial", 12)
)
name_entry.pack(pady=5)


# Roll Number
tk.Label(
    root,
    text="Roll Number",
    font=("Arial", 12)
).pack()

roll_entry = tk.Entry(
    root,
    width=40,
    font=("Arial", 12)
)
roll_entry.pack(pady=5)


# Marks
tk.Label(
    root,
    text="Marks",
    font=("Arial", 12)
).pack()

marks_entry = tk.Entry(
    root,
    width=40,
    font=("Arial", 12)
)
marks_entry.pack(pady=5)


# Add Button
tk.Button(
    root,
    text="Add Student",
    command=add_student,
    width=15
).pack(pady=10)


# Display students
display_list = tk.Listbox(
    root,
    width=65,
    height=10,
    font=("Arial", 12)
)
display_list.pack(pady=10)


# Buttons
button_frame = tk.Frame(root)
button_frame.pack(pady=10)

tk.Button(
    button_frame,
    text="Delete",
    command=delete_student,
    width=12
).grid(row=0, column=0, padx=5)

tk.Button(
    button_frame,
    text="Sort",
    command=sort_marks,
    width=12
).grid(row=0, column=1, padx=5)

tk.Button(
    button_frame,
    text="Analyze",
    command=analyze_result,
    width=12
).grid(row=0, column=2, padx=5)

tk.Button(
    button_frame,
    text="Clear",
    command=clear_data,
    width=12
).grid(row=0, column=3, padx=5)


# Search
search_frame = tk.Frame(root)
search_frame.pack(pady=10)

tk.Label(
    search_frame,
    text="Search Roll No."
).grid(row=0, column=0, padx=5)

search_entry = tk.Entry(
    search_frame,
    width=20
)
search_entry.grid(row=0, column=1, padx=5)

tk.Button(
    search_frame,
    text="Search",
    command=search_student
).grid(row=0, column=2, padx=5)


# Result analysis
result_label = tk.Label(
    root,
    text="Total:     Average:     Maximum:     Minimum:",
    font=("Arial", 12, "bold")
)
result_label.pack(pady=15)


# Run application
root.mainloop()
```
