```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
xaxis= np.array([23,44,55,66])
yaxis = np.array([100,205,400,18])
```

```python
plt.figure(figsize=(7,7),facecolor='pink')
sns.set_style('darkgrid')
sns.set_theme('notebook')
sns.pointplot(yaxis,ms=20,mec='m',mfc='g',color='pink')
```

```python
plt.figure(figsize=(7,7),facecolor="#E1FF00")
plt.xlabel('xaxis')
plt.ylabel('yaxis')
plt.title('graph of xaxis or yaxis')
sns.set_style('darkgrid')
sns.set_theme('notebook')
plt.plot(yaxis,xaxis,'-o',ms=20,mec='m',mfc='g',color="#0059FF")
plt.subplot(2,2,2)
plt.plot(yaxis,xaxis,'-o',ms=20,mec='m',mfc='g',color="#0059FF")
plt.xlabel('xaxis')
plt.ylabel('yaxis')
plt.grid(axis='x',color="#009DFF")
```
