> ## *****create the scatter plot graph of cgpa  is less than <7 in x1 and y1 or cgpa is greatoe than and equal to 7 in x2 and y2 and create the scatter plot*****
* c : red and blue
* func : plt.scatter(),plt.lagend()

```python
# Select students having CGPA less than 7
x1 = np.array(df[df['cgpa'] < 7]['cgpa'])
y1 = np.array(df[df['cgpa'] < 7]['package'])

# Select students having CGPA greater than or equal to 7
x2 = np.array(df[df['cgpa'] >= 7]['cgpa'])
y2 = np.array(df[df['cgpa'] >= 7]['package'])

# Display the first scatter plot in red
plt.scatter(x1, y1, c="red", label="CGPA Below 7")

# Display the second scatter plot in blue
plt.scatter(x2, y2, c="blue", label="CGPA 7 or Above")

# Add a label to the x-axis
plt.xlabel("CGPA")

# Add a label to the y-axis
plt.ylabel("Package")

# Add a title to the graph
plt.title("Comparison of Student Placement Packages")

# Display the legend
plt.legend()

# Display the graph
plt.show()
```
