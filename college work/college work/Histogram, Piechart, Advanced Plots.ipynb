Skewness measures the asymmetry of a data distribution.
Skewness is calculated using the third standardized moment of a dataset. It measures how far a data distribution is tilted or asymmetric.
1. Positive Skewness (Right Skew)
In a positively skewed distribution, the right tail is longer than the left which means most data points are on the left with a few large values pulling the distribution to the right.

Relationship:

Mean > Median > Mode

Examples: Income distribution, exam scores and stock market returns.

2. Negative Skewness (Left Skew)
In a negatively skewed distribution, the left tail is longer which means most data points are on the right with a few smaller values pulling the distribution to the left.

Relationship:

Mean < Median < Mode

Examples: Test scores on easy exams, age at retirement and gestational age at birth.

3. Zero Skewness (Symmetrical Distribution)
Zero skewness shows a perfectly symmetrical distribution where the mean, median and mode are equal. In a symmetrical distribution, the data points are evenly distributed around the central point.

Relationship:

Mean = Median = Mode

Example: A perfectly balanced dataset with equal frequencies of all values.



```python
# Calculate skewness
skewvalue = df['selling_price'].skew()

# Create histogram
plt.hist(df['selling_price'], color='skyblue', edgecolor='black')

# Add skewness value on graph
plt.text(0.65, 0.90, "Skewness = " + str(round(skewvalue, 2)),
         transform=plt.gca().transAxes)

# Add labels and title
plt.xlabel("Selling Price")
plt.ylabel("Frequency")
plt.title("Histogram Showing Skewness")

# Display graph
plt.show()
```

HeatMap:
A heatmap is a graphical representation of numerical data in which different colours represent different values.

It is commonly used to visualize the correlation between numerical variables.

Seaborn Library needed for heat map

```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate correlation
corrdata = df[['km_driven', 'selling_price']].corr()

# Create heatmap
plt.figure(figsize=(7, 5))
sns.heatmap(corrdata, annot=True, cmap='coolwarm',fmt='.2f', linewidths=1) 

#fmt=format
#annot means annotation. It is used to display numerical values inside the heatmap cells.
#cmap means colormap:cmap='viridis',cmap='plasma',cmap='inferno',cmap='Blues',cmap='Reds'

# Add title
plt.title("Correlation Heatmap of Cars Dataset")

# Display graph
plt.show()
```

```python
# Create heatmap with cell borders
sns.heatmap(corrdata, annot=True, linewidths=2)

# Display graph
plt.show()
```

vmin → Minimum colour scale value.
vmax → Maximum colour scale value.

```python
# Create heatmap with minimum and maximum colour values
sns.heatmap(corrdata, annot=True, cmap='coolwarm',
            vmin=-1, vmax=1)

# Add title
plt.title("Correlation Heatmap of Cars Dataset")

# Display graph
plt.show()
```

Cluster Map: A cluster map is a visualization that combines a heatmap with hierarchical clustering.

```python
# Create cluster map
sns.clustermap(df.groupby('brand')[['km_driven', 'selling_price']].mean(),
               cmap='viridis')

# Display graph
plt.show()
```

1. What is z_score?
z_score standardizes the data so that features with different scales can be compared more fairly.

It has two common options:
- z_score=0 – Standardizes each column (feature).
- z_score=1 – Standardizes each row.

```python
# Calculate average values for each brand
branddata = df.groupby('brand')[['km_driven', 'selling_price']].mean()

# Create cluster map with z_score
sns.clustermap(branddata, cmap='coolwarm', z_score=0)

# Display graph
plt.show()
```

```python
# Calculate average values for each brand
branddata = df.groupby('brand')[['km_driven', 'selling_price']].mean()

# Create cluster map using average method
sns.clustermap(branddata, cmap='viridis',
               z_score=0, method='average')

# Display graph
plt.show()
```

```python
# Calculate average values for each brand
branddata = df.groupby('brand')[['km_driven', 'selling_price']].mean()

# Create cluster map
g = sns.clustermap(branddata, cmap='viridis',z_score=0, method='average')

# Add title
g.figure.suptitle("Car Brand Clustering", y=1.02)

# Display graph
plt.show()
```

Scatter Plot – plt.scatter()
Definition: A scatter plot is used to show the relationship between two numerical variables. Each dot represents one observation.

```python
# Create scatter plot
plt.scatter(df['km_driven'], df['selling_price'])

# Add title and labels
plt.title("Kilometres Driven vs Selling Price")
plt.xlabel("Kilometres Driven")
plt.ylabel("Selling Price")

# Display graph
plt.show()
```

s – Size of points
s=20 → Small points
s=50 → Medium-sized points
s=100 → Large points
s=200 → Very large points

alpha – Transparency of points
The value of alpha ranges from 0 to 1.
alpha=0 → Completely transparent (invisible)
alpha=0.3 → More transparent
alpha=1 → Completely opaque (no transparency)

```python
# Create scatter plot with customization
plt.scatter(df['km_driven'], df['selling_price'],
            color='red', s=50, alpha=0.6,
            marker='o', edgecolor='black')

# Add title and labels
plt.title("Kilometres Driven vs Selling Price")
plt.xlabel("Kilometres Driven")
plt.ylabel("Selling Price")

# Display graph
plt.show()
```

```python
# Create advanced scatter plot
plt.scatter(df['km_driven'], df['selling_price'],
            c=df['selling_price'], cmap='coolwarm',
            vmin=0, vmax=1000000,
            s=80, alpha=0.6,
            edgecolors='black', linewidths=0.5)

# Add colour bar
plt.colorbar(label='Selling Price')

# Add title and labels
plt.title("Advanced Scatter Plot")
plt.xlabel("Kilometres Driven")
plt.ylabel("Selling Price")

# Display graph
plt.show()
```
