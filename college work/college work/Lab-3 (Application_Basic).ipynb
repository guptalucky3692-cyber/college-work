```python
# Calculate Discounted Price of a Product
price = float(input("Enter product price: "))
discount = float(input("Enter discount percentage: "))

discount_amount = price * discount / 100
discounted_price = price - discount_amount

print("Discount Amount:", discount_amount)
print("Discounted Price:", discounted_price)

```

```python

# Calculate GST on a Product
price = float(input("Enter product price: "))
gst = float(input("Enter GST percentage: "))

gst_amount = price * gst / 100
final_price = price + gst_amount

print("GST Amount:", gst_amount)
print("Final Price:", final_price)
```

```python
# Calculate Final Bill After Discount and GST
price = float(input("Enter product price: "))
discount = float(input("Enter discount percentage: "))
gst = float(input("Enter GST percentage: "))

discount_amount = price * discount / 100
discounted_price = price - discount_amount
gst_amount = discounted_price * gst / 100
final_bill = discounted_price + gst_amount

print("Discount Amount:", discount_amount)
print("GST Amount:", gst_amount)
print("Final Bill:", final_bill)
```

```python
# Calculate Profit or Loss
cost_price = float(input("Enter cost price: "))
selling_price = float(input("Enter selling price: "))

if selling_price > cost_price:
    profit = selling_price - cost_price
    print("Profit:", profit)
elif cost_price > selling_price:
    loss = cost_price - selling_price
    print("Loss:", loss)
else:
    print("No Profit No Loss")
```

```python

# Calculate Electricity Bill
units = float(input("Enter electricity units: "))

if units <= 100:
    bill = units * 3
elif units <= 200:
    bill = (100 * 3) + ((units - 100) * 5)
else:
    bill = (100 * 3) + (100 * 5) + ((units - 200) * 7)

print("Electricity Bill:", bill)
```
