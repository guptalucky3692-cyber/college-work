```python
#for loop
for i in range(1, 11):
    print(i, end=" ")
    
#User Defined for loop
def numbers():
    for i in range(1, 101):
        print(i)
numbers()

#Print numbers from 10 to 1
for i in range(10, 0, -1):
    print(i)
    
#Print even numbers from 1 to 20
for i in range(2, 21, 2):
    print(i, end=" ")
    
# Print odd numbers from 1 to 20
for i in range(1, 21, 2):
    print(i, end=" ")
    
#Print multiplication table
n = int(input("Enter a number: "))

for i in range(1, 11):
    print(n * i, end=" ")
    
#User defined multiplication table
n = int(input("Enter a number: "))
for i in range(1, 11):
    print(n, "x", i, "=", n * i)
    
#Find factorial of a number
n = int(input("Enter a number: "))

fact = 1
for i in range(1, n + 1):
    fact = fact * i

print("Factorial =", fact)

#Print prime numbers from 1 to 100
for num in range(1, 101):   
    if num > 1:
        for i in range(2, num):
            if (num % i) == 0:
                break
        else:
            print(num)
```

```python
# Simple Calculator Using Menu
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))

print("Addition")
print("Subtraction")
print("Multiplication")
print("Division")

choice = int(input("Enter your choice: "))

if choice == 1:
    print("Result =", a + b)

elif choice == 2:
    print("Result =", a - b)

elif choice == 3:
    print("Result =", a * b)

elif choice == 4:
    print("Result =", a / b)

else:
    print("Invalid choice")
```

String indexing

```python

s = "Welcome to Python Programming. Python is a powerful and easy-to-learn programming language. It is widely used in Data Science, Artificial Intelligence, Machine Learning, and Web Development."

# Display the complete string
print(s)

# Find the length of the string
print(len(s))

# Access the first character
print(s[0])

# Access the last character
print(s[-1])

# Access the character at index 10
print(s[10])

# Print the first 10 characters
print(s[:10])

# Print the last 10 characters
print(s[-10:])

# Print characters from index 10 to 30
print(s[10:31])

# Print the complete string using slicing
print(s[:])

# Print every second character
print(s[::2])

# Print characters at odd indexes
print(s[1::2])

# Reverse the string
print(s[::-1])

# Remove the first character
print(s[1:])

# Remove the last character
print(s[:-1])

# Convert the string to uppercase
print(s.upper())

# Convert the string to lowercase
print(s.lower())

# Capitalize the first letter
print(s.capitalize())

# Convert each word's first letter to uppercase
print(s.title())

# Count occurrences of the word Python
print(s.count("Python"))

# Find the position of Programming
print(s.find("Programming"))

# Replace Python with Java
print(s.replace("Python", "Java"))

# Check whether Python exists in the string
print("Python" in s)

# Print each character using indexing
for i in range(len(s)):
    print(i, s[i])
```
